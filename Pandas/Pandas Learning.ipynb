# Handling Missing values


```python
data = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [1, 2, 3, 4, 5],
    'C': [1, 2, 3, np.nan, np.nan],
    'D': [1, np.nan, np.nan, np.nan, 5]
}
df = pd.DataFrame(data)
df
```

```python
# to find which cell is NA 
df.isna()
df.isna().sum() # to find which column has how many NA values
df.isna().any() # to find if any column has any NA values?
df.dropna() # this will remove the rows with NA values
# dropna() works in row wise
df.dropna(thresh=3) # thresh is for checking id it has that number of nonNA values or not


df.fillna('pong') # to fill the NA values with the set number/string.
# fillna() works in column wise
vals={'A':-1,'B':-2,"C":-3,"D":-4}
df.fillna(value=vals)
df.fillna(df.mean()) # fill with mean value of that column
```

# Merging, joining , Concatination

```python
employees = pd.DataFrame({
    'employee_id': [1, 2, 3, 4, 5],
    'name': ['John', 'Anna', 'Peter', 'Linda', 'Bob'],
    'department': ['HR', 'IT', 'Finance', 'IT', 'HR']
})

# DataFrame 2: Salary information
salaries = pd.DataFrame({
    'employee_id': [1, 2, 3, 6, 7],
    'salary': [60000, 80000, 65000, 70000, 90000],
    'bonus': [5000, 10000, 7000, 8000, 12000]
})
```

```python
print(employees)
print(salaries)

# merging 
pd.merge(employees,salaries) # merging two tables 
pd.merge(employees,salaries,on='employee_id') # setting on which column i need the merging 
pd.merge(employees,salaries,on='employee_id',how='outer') # outer join
pd.merge(employees,salaries,on='employee_id',how='right') # right join
```

# Group By and Aggregation 

```python
data = {
    'Category': ['A', 'B', 'A', 'B', 'A', 'B', 'A', 'B'],
    'Store': ['S1', 'S1', 'S2', 'S2', 'S1', 'S2', 'S2', 'S1'],
    'Sales': [100, 200, 150, 250, 120, 180, 200, 300],
    'Quantity': [10, 15, 12, 18, 8, 20, 15, 25],
    'Date': pd.date_range('2023-01-01', periods=8)
}
df = pd.DataFrame(data)
df 
```

```python
cat=df.groupby(['Store','Category'])[['Sales','Quantity']].sum()
print(cat)
```

## Aggregation
- multiple things combined and then presented

```python
cat=df.groupby(['Store','Category'])[['Sales','Quantity']].sum()
cat['Mean']=df.groupby(['Store','Category'])[['Sales']].mean()
cat['std']=df.groupby(['Store','Category'])[['Sales']].std()

# instead of doing the statistical oparation manually we will use aggregation 
new_cat=cat['Sales'].agg(['sum','mean','median','max','min','count','std'])
bat=df['Sales'].agg(['sum','mean','median','max','min','count','std'])
print(cat)
print(new_cat)
print(bat)
```

# Feature Extraction

- [X] make a new column for episode count
- [X] make a new column for time stamp
- [X] which anime has the highest score
- [X] give me top 5 highest scoring anime
- [X] which anime has the highest episode count
- [X] animes with top 5 episode count
- [X] which is the longest running anime

```python
df=pd.read_csv('/kaggle/input/datasets/abhigangwar30/anime-project/anime.csv')
df.head(5)
```

```python
df.loc[0]
```

```python
# since the episode count is inside title bounded by () then i need to extract that using a function
def exct_epi(title):
    if '(' in title:
        start=title.index('(')
        if ')' in title and title.index(')')>start:
            epi_count=title[start+1:title.index(')')]
            epi_count=epi_count.split(' ')
            return int(epi_count[0])

print(exct_epi(df.loc[2]['Title']))

# extract time stamp from the title
def exct_ts(title):
    start=title.index(')')+1
    #print(start)
    ts=title[start:start+19]
    return ts

print(exct_ts(df.loc[0]['Title']))

# extract no of monthes the show is aired
mon={'Jan':1,'Feb':2,'Mar':3,'Apr':4,'May':5,'Jun':6,'Jul':7,'Aug':8,'Sep':9,'Oct':10,'Nov':11,'Dec':12}
def mon_cnt(colm):
    colm=colm.split(' - ')
    m1,y1=colm[0].split(' ')
    m2,y2=colm[1].split(' ')
    cnt=0
    if (int(y1)<int(y2)):
        cnt=int(mon[m2])+(12-int(mon[m1])+1)
    elif (int(y1)==int(y2)):
        cnt=int(mon[m2])-int(mon[m1])+1
    else:
        return "NULL"
    return cnt

print(mon_cnt(exct_ts(df.loc[0]['Title'])))
        
```

```python
df["EPISODES"]=[exct_epi(df.loc[title]['Title']) for title in range(int(df.shape[0]))]
df['Time-Stamp']=df['Title'].apply(exct_ts)
df['Months-Aired']=df['Time-Stamp'].apply(mon_cnt)
df
```

```python
# which anime has the highest score 
title=df[df['Score'] == df['Score'].max()]['Title']
title

```

```python
#top 5 highest scoring anime
df['Title'].head(5)
```

```python
# which anime has the highest episode count
df[df['EPISODES']== df['EPISODES'].max()][['Title','EPISODES']]
```

```python
# animes with top 5 episode count
df.sort_values(by='EPISODES',ascending=False)['Title'].head(5)
```

```python
#which is the longest running anime
df[df['Months-Aired']== df['Months-Aired'].max()]
#df.sort_values(by='Months-Aired',ascending=False)
```
